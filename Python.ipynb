```python
# Sformatuj metryki modelu.
# Wymagania:
# Wyświetl każdą metrykę w procentach z 2 miejscami po przecinku
# Użyj f-strings z formatowaniem :.2%
# Dodaj estetyczne wyrównanie kolum
Accuracy = 0.9234
Precision = 0.8891
Recall = 0.9456

print(f"Accuracy: {Accuracy:.2f}")
print(f"Precision: {Precision:.2f}")
print(f"Recall: {Recall:.2f}")

print(f"Accuracy: {Accuracy:.2%}")
print(f"Precision: {Precision:.2%}")
print(f"Recall: {Recall:.2%}")

headers = ["Accuracy", "Precision", "Recall"]
print(f"{headers[0]:<10} {headers[1]:>10} {headers[2]:>10}")
```

```python
# Znormalizuj wartość do zakresu [0, 1].
# Wymagania:
# Wzór: normalized = (value - min) / (max - min)
# Oblicz dla wartości 75, 50, 100, 62.5
# Sprawdź, czy wynik mieści się w [0, 1]
# Wyświetl tabelkę: wartość oryginalna | wartość znormalizowan
import math

Wartość = 50 # /Oblicz dla wartości 75, 50, 100, 62.5
Min = 50
Max = 100

Normalized = (Wartość - Min)/(Max - Min)
print(Normalized)

if 0 <= Normalized <=1:
    print("Wynik mieści się w zakresie [0,1]")
else:
    print("Wynik nie mieści się w zakresie [0,1]")

print(f"\nPomiary:")
print(f"Wartość: {Wartość}")
print(f"Wartość znormalizowana: {Normalized}")
```

```python
# Napisz program, który obliczy wskaźnik BMI (Body Mass Index).
# Wymagania:
# Stwórz zmienne waga i wzrost
# Oblicz BMI według wzoru: BMI = waga / (wzrost²)
# Wyświetl wynik z 2 miejscami po przecinku
# Dodaj interpretację: BMI < 18.5 (niedowaga), 18.5-24.9 (norma), ≥25 (nadwaga)

waga = 70
wzrost = 1.75
BMI = waga / (wzrost**2)
print(f'BMI {BMI:.2f}') # 2 miejsca po przecinku
if BMI < 18.5:
    print('Niedowaga')
elif BMI >= 18.5:
    print('Norma')
elif BMI >= 25:
    print('Nadwaga')

# # elif BMI > 25
# #     print('Norma')
# # else:
# #     print('Nadwaga')
```

```python
# Napisz program konwertujący temperatury Celsius na Fahrenheit i Kelvin.
# Dane:
#   Temperatura: 23.5°C
# Wymagania:
# - Fahrenheit = Celsius × 9/5 + 32
# - Kelvin = Celsius + 273.15
# - Wyświetl wszystkie 3 wartości sformatowane z 1 miejscem po przecink
# Oczekiwany rezultat
#   23.5°C = 74.3°F = 296.6 K
temp = 23.5
far = temp * (9/5) + 32
Kelvin = temp + 273.15

print (f'far) {far:.1f}')
print (f'Kelvin {Kelvin:.1f}')
```
