```python
import sys
import os
# Sprawdzamy, czy pracujemy w wirtualnym środowisku
print("=== ŚRODOWISKO PYTHON ===")
print(f"Python executable: {sys.executable}")
print(f"Python version: {sys.version}")
# Jeśli ścieżka zawiera 'ds_env' lub inną nazwę venv - jesteś w środowisku!
if 'ds_env' in sys.executable or 'venv' in sys.executable:
    print("\n✓ Pracujesz w wirtualnym środowisku!")
else:
    print("\n⚠ UWAGA: Prawdopodobnie używasz systemowego Pythona")
# Lista zainstalowanych pakietów
import pkg_resources
installed_packages = [d.project_name for d in pkg_resources.working_set]
print(f"\nZainstalowane pakiety: {len(installed_packages)}")
print("Przykładowe pakiety:")
for pkg in sorted(installed_packages)[:10]:
print(f" - {pkg}")
```
