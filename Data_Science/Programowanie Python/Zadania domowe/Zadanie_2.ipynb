```python
# Masz string CSV: "Alice,28,Data Scientist,75000" . Użyj split(',') aby wyciągnąć:
# Imię
# Wiek (jako int)
# Zawód
# Pensję (jako int)
tekst = "Alice,28,Data Scientist,75000"
tekst_1 = tekst[0:5]
print(f"Imię: {tekst_1}")
tekst_2 = tekst[6:8]
print(f"Wiek: {tekst_2}")
tekst_3 = tekst[9:23]
print(f"Zawód: {tekst_3}")
tekst_4 = tekst[24:29]
print(f"Pensja: {tekst_4}")

tekst_nowy = tekst.replace(",", " ")
print(tekst_nowy)
```

```python
# Masz listę pomiarów: [23, 24, 22, 150, 23, 25, -10, 24, 23] . Outlierem jest wartość
# poza zakresem [20, 30]. Stwórz nową listę bez outlierów używając list comprehension.

pomiary = [23, 24, 22, 150, 23, 25, -10, 24, 23]
print(f"Lista pomiarów: {pomiary}")
wynik = [x for x in pomiary if 20 <= x <= 30]
print(f"Pomiary poza zakresem: {wynik}")
for x in wynik:
    pomiary.remove(x)
print(f"Nowa lista: {pomiary}")
```
