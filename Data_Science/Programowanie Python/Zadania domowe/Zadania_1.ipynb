```python
# Ustandaryzuj nazwę kolumny z datasetu.
# Wymagania:
# Usuń białe znaki
# Zamień na małe litery
# Zastąp _ przez " "
# Użyj title case
Nazwa_raw = " PATIENT_age_YEARS "
Nazwa_clean = Nazwa_raw.strip()
print(f"Orginał: '{Nazwa_raw}'")
print(f"Po strip(): '{Nazwa_clean}'")

print(Nazwa_clean.lower())

Nazwa_standard = Nazwa_clean.replace("_", " ")
print(f"Standaryzowana: '{Nazwa_standard}'")

print(Nazwa_standard.title())
```
