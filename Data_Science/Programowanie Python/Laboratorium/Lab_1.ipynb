```python
# Napisz program, który obliczy wskaźnik BMI (Body Mass Index).
# Wymagania:
# Stwórz zmienne waga i wzrost
# Oblicz BMI według wzoru: BMI = waga / (wzrost²)
# Wyświetl wynik z 2 miejscami po przecinku
# Dodaj interpretację: BMI < 18.5 (niedowaga), 18.5-24.9 (norma), ≥25 (nadwaga)

waga = 70
wzrost = 1.75
BMI = waga / (wzrost**2)
print(f'BMI {BMI:.2f}') # 2 miejsca po przecinku
if BMI < 18.5:
    print('Niedowaga')
elif BMI >= 18.5:
    print('Norma')
elif BMI >= 25:
    print('Nadwaga')

# # elif BMI > 25
# #     print('Norma')
# # else:
# #     print('Nadwaga')
```

```python
# Napisz program konwertujący temperatury Celsius na Fahrenheit i Kelvin.
# Dane:
#   Temperatura: 23.5°C
# Wymagania:
# - Fahrenheit = Celsius × 9/5 + 32
# - Kelvin = Celsius + 273.15
# - Wyświetl wszystkie 3 wartości sformatowane z 1 miejscem po przecink
# Oczekiwany rezultat
#   23.5°C = 74.3°F = 296.6 K
temp = 23.5
far = temp * (9/5) + 32
Kelvin = temp + 273.15

print (f'far) {far:.1f}')
print (f'Kelvin {Kelvin:.1f}')
```
