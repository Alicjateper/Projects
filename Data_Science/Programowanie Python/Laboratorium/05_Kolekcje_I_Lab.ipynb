```python
# Temperaturę z pierwszego dnia
# Temperaturę z ostatniego dnia
# Temperatury z trzech środkowych dni (slicing)
temp = [22.5, 23.0, 21.5, 24.0, 23.5,
22.0, 23.0]

print(f"Pierwszy dzień, {temp[0]}")
print(f"Ostatni dzień, {temp[-1]}")
print(f"Temperatura trzech środkowych temp, {temp[2:5]}")
```

```python
results = []
accuracy = [0.75, 0.82, 0.79, 0.85, 0.88]
results.extend(accuracy)
# Wyświetl długość listy
# Wyświetl średnią wartość
# Wyświetl maksymalną wartość
print(f"Dodanie do listy, {results} ")

resualts2 = len(results)
print(f"Długość listy wynosi: {resualts2}")
średnia = sum(results) / resualts2
print(f"Średnia wynosi: {średnia}")
max = max(results)
print(f"Maxymalna watrość wynosi: {max}")
```

```python
# Masz listę wyników: [0.78, 0.92, 0.85, 0.88, 0.81] . Posortuj ją malejąco i wyświetl 3
# najlepsze wyniki.
lista = [0.78, 0.92, 0.85, 0.88, 0.81]
lista.sort(reverse=True)
print(f"malejąco: {lista}")
top = lista[:3]
print(f"Najlepsze wyniki to: {top}")
```
