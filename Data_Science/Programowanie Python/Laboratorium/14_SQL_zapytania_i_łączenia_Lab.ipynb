```python
# Utwórz dwie tabele: customers (customer_id, name, city) i orders (order_id,
# customer_id, amount).
# Wymagania:
# Wykonaj INNER JOIN łączący klientów z zamówieniami
# Wyświetl nazwę klienta, miasto i kwotę zamówienia
# Posortuj wyniki według kwoty malejąco

import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')

customers = pd.DataFrame({ 
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Jan Kowalski', 'Piotr Nowak', 'Anna Kot', 'Maria Nowak', 'Jan Kot'],
    'city': ['Rzeszów', 'Kraków', 'Warszawa', 'Łódź', 'Lublin']
})

orders = pd.DataFrame({
    'order_id': ['Z1', 'Z2' ,'Z3' ,'Z4', 'Z5'],
    'customer_id': [5, 5, 1, 4, 2],
    'amount': [350, 1500, 5000, 200, 845]
})

customers.to_sql('customers', conn, index=False)
orders.to_sql('orders', conn, index=False)

# Wykonaj INNER JOIN łączący klientów z zamówieniami
query = """SELECT 
            c.customer_id,
            c.name,
            c.city,
            o.order_id,
            o.amount
            FROM customers c
            INNER JOIN orders o ON c.customer_id = o.customer_id
            ORDER BY o.amount """
print('Klienci i ich zamówienia:\n', pd.read_sql_query(query,conn))
```

```python
# Wymagania:
# Używając danych sprzedażowych, pogrupuj według kategorii
# Użyj HAVING aby znaleźć tylko kategorie z przychodem > 1000 PLN
# Oblicz liczbę transakcji, łączny przychód i średnią dla każdej kategorii
# Posortuj według łącznego przychodu

import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')
# dane pobrane z notatek

sales = pd.DataFrame({
'sale_id': range(1, 21),
'product': ['Laptop', 'Mysz', 'Klawiatura', 'Monitor', 'Słuchawki'] * 4,
'category': (['Elektronika'] * 2 + ['Akcesoria'] * 3) * 4,
'revenue': [3500, 50, 200, 1200, 150,
3600, 45, 210, 1250, 140,
3400, 55, 190, 1180, 160,
3550, 48, 205, 1220, 155]
})

sales.to_sql('sales', conn, index=False)

# Użyj HAVING aby znaleźć tylko kategorie z przychodem > 1000 PLN

query = """SELECT
            category,
            COUNT(*) AS total_count,
            SUM(revenue) AS total_revenue,
            AVG(revenue) AS avg_revenue
            FROM sales
            GROUP BY category
            HAVING SUM(revenue) > 1000
            ORDER BY total_revenue """

print('Wynik:\n', pd.read_sql_query(query,conn))
```

```python
# Dataset: Titanic
# Wymagania:
# Znajdź pasażerów 1 klasy, kobiety, które przeżyły
# Oblicz średni wiek i średnią cenę biletu dla tej grupy
# Porównaj z mężczyznami z 3 klasy którzy nie przeżyli
# Stwórz wykres porównawczy (barplot)

import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect(':memory:')

plik = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

titanic = pd.read_csv(plik)
titanic.head()
titanic.columns

titanic.to_sql('titanic', conn, index=False)

# Znajdź pasażerów 1 klasy, kobiety, które przeżyły

query = """SELECT
        COUNT(*) AS Survived
        from titanic
        WHERE Survived = 1 AND Sex = 'female' AND Pclass = 1"""
print('Ilość kobiet z 1 klasy, które przeżyły:\n', pd.read_sql_query(query,conn))

# Oblicz średni wiek i średnią cenę biletu dla tej grupy
query1 = """SELECT
        ROUND(AVG(Age), 0) AS avg_age,
        ROUND(AVG(Fare), 0) AS avg_fare
        from titanic
        WHERE Survived = 1 AND Sex = 'female' AND Pclass = 1 
        """
print('Średni wiek i średnia cena biletu dla kobiet z 1 klasy:\n', pd.read_sql_query(query1,conn))

# Porównaj z mężczyznami z 3 klasy którzy nie przeżyli
query2 = query2 = """SELECT
        ROUND(AVG(Age), 0) AS avg_age,
        ROUND(AVG(Fare), 0) AS avg_fare
        from titanic
        WHERE Survived = 0 AND Sex = 'male' AND Pclass = 3 """
print('Ilość mężczyzn z 3 klasy, którzy nieprzeżyły:\n', pd.read_sql_query(query2,conn))

# dane do wykresu AI
women_stats = pd.read_sql_query(query1, conn)
men_stats = pd.read_sql_query(query2, conn)

labels = ['Kobiety 1 klasa (przeżyły)', 'Mężczyźni 3 klasa (nie przeżyli)']

avg_age = [women_stats['avg_age'][0], men_stats['avg_age'][0]]
avg_fare = [women_stats['avg_fare'][0], men_stats['avg_fare'][0]]

import matplotlib.pyplot as plt

x = range(len(labels))

plt.figure(figsize=(10,5))

plt.bar(x, avg_age, width=0.4, label='Średni wiek')
plt.bar([i + 0.4 for i in x], avg_fare, width=0.4, label='Średnia cena biletu')

plt.xticks([i + 0.2 for i in x], labels, rotation=20)
plt.ylabel('Wartość')
plt.title('Porównanie grup pasażerów Titanic')
plt.legend()

plt.show()
```
