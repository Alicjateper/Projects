```python
# Napisz program, który obliczy wskaźnik BMI (Body Mass Index).
# Wymagania:
# Stwórz zmienne waga i wzrost
# Oblicz BMI według wzoru: BMI = waga / (wzrost²)
# Wyświetl wynik z 2 miejscami po przecinku
# Dodaj interpretację: BMI < 18.5 (niedowaga), 18.5-24.9 (norma), ≥25 (nadwaga)

waga = 70
wzrost = 1.75
BMI = waga / (wzrost**2)
print(f'BMI {BMI:.2f}') # 2 miejsca po przecinku
if BMI < 18.5:
    print('Niedowaga')
elif BMI >= 18.5:
    print('Norma')
elif BMI >= 25:
    print('Nadwaga')

# # elif BMI > 25
# #     print('Norma')
# # else:
# #     print('Nadwaga')
```

```python
# Napisz program konwertujący temperatury Celsius na Fahrenheit i Kelvin.
# Dane:
#   Temperatura: 23.5°C
# Wymagania:
# - Fahrenheit = Celsius × 9/5 + 32
# - Kelvin = Celsius + 273.15
# - Wyświetl wszystkie 3 wartości sformatowane z 1 miejscem po przecink
# Oczekiwany rezultat
#   23.5°C = 74.3°F = 296.6 K
temp = 23.5
print(f"Temperatura =: {temp}")
Fah = temp * (9/5) + 32
print(f"Fahrenheit =: {Fah:.1f}")
Kel = temp + 273.15
print(f"Kelvin =: {Kel:.1f}")
```

```python
# Oblicz rozmiary zbiorów treningowego, walidacyjnego i testowego.
# Dane: Całkowita liczba próbek: 1000
# Proporcje: 70% trening, 15% walidacja, 15% test
# Wymagania:
# Użyj mnożenia i dzielenia całkowitego
# Upewnij się, że suma = 1000 (obsłuż resztę z dzielenia)
# Wyświetl rozmiary wszystkich zbiorów

L_próbek = 1000
trening = L_próbek * 70 // 100
print(f"Trening wynosi {trening}")
walidacja = L_próbek * 15 // 100
print(f"Walidacja wynosi {walidacja}")
test = L_próbek - walidacja - trening
print(f"Test wynosi {test}")
```

```python
# Oblicz learning rate dla schedulera.
# Dane:
# Początkowy learning rate: 0.1
# Współczynnik decay: 0.95
# Numer epoki: 10
# Wymagania:
# Użyj wzoru: lr_new = lr_initial × (decay ^ epoch)
# Operator potęgowania **
# Wyświetl wynik w notacji naukowej ( .2e )
learning_rate = 0.1
decay = 0.95
epoch = 10
lr_new = learning_rate * (decay ** epoch)
print(f"Learning rate po 10 epokach wynosi {lr_new:.2e}")
```
