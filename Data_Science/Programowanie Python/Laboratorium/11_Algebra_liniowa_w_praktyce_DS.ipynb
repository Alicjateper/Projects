```python
# Stwórz dwa wektory 2D: v1 = [3, 4] i v2 = [1, 2]. Oblicz ich sumę, różnicę oraz narysuj
# wszystkie trzy wektory na wykresie używając plt.quiver() .
# Stwórz wektory jako NumPy arrays
# Oblicz v1 + v2 i v1 - v2
# Narysuj wszystkie 4 wektory (v1, v2, suma, różnica) różnymi kolorami
# Dodaj legendę i grid
# Wykres z 4 strzałkami reprezentującymi wektory

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 8))

origin = np.array([0, 0])

# Stwórz wektory jako NumPy arrays
v1 = np.array([3,4])
print("Wektor v1:\n", v1)

v2 = np.array([1,2])
print("Wektor v2:\n", v2)

# Oblicz v1 + v2 i v1 - v2
suma = v1 + v2
print("Suma wektorów wynosi:\n", suma)

roznica = v1 - v2
print("Różnica wektorów wynosi:\n", roznica)

# Narysuj wszystkie 4 wektory (v1, v2, suma, różnica) różnymi kolorami
ax.quiver(*origin, *v1, angles='xy', scale_units='xy', scale=1, color='blue', width=0.006, label='v1 = [3, 4]')
ax.quiver(*origin, *v2, angles='xy', scale_units='xy', scale=1, color='red', width=0.006, label='v2 = [1, 2]')
ax.quiver(*origin, *suma, angles='xy', scale_units='xy', scale=1, color='green', width=0.006, label='v1 + v2 = [4, 6]')
ax.quiver(*origin, *roznica, angles='xy', scale_units='xy', scale=1, color='pink', width=0.006,label='v1 - v2 = [2, 2]')

# Równoległobok dodawania wektorów (wizualizacja sumy)
ax.quiver(*v1, *v2, angles='xy', scale_units='xy', scale=1, color='red', width=0.004, alpha=0.3, linestyle='dashed')
ax.quiver(*v2, *v1, angles='xy', scale_units='xy', scale=1, color='blue', width=0.004, alpha=0.3, linestyle='dashed')

ax.set_xlim(-1, 10)
ax.set_ylim(-1, 9)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)

ax.set_title('Dodawanie wektorów w przestrzeni 2D', fontsize=14)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```
