```python
results = []
accuracy = [0.75, 0.82, 0.79, 0.85, 0.88]
results.extend(accuracy)
# Wyświetl długość listy
# Wyświetl średnią wartość
# Wyświetl maksymalną wartość
print(f"Dodanie do listy, {results} ")

resualts2 = len(results)
print(f"Długość listy wynosi: {resualts2}")
średnia = sum(results) / resualts2
print(f"Średnia wynosi: {średnia}")
max = max(results)
print(f"Maxymalna watrość wynosi: {max}")
```

```python
# Masz listę wyników: [0.78, 0.92, 0.85, 0.88, 0.81] . Posortuj ją malejąco i wyświetl 3
# najlepsze wyniki.
lista = [0.78, 0.92, 0.85, 0.88, 0.81]
lista.sort(reverse=True)
print(f"malejąco: {lista}")
top = lista[:3]
print(f"Najlepsze wyniki to: {top}")
```

```python
text = [" Customer ID ", "TOTAL AMOUNT", " date "]
# Usuń białe znaki (strip)
delete = []
for t in text:
    t = t.strip()
    delete.append(t)
print(delete)
# Zamień na małe litery (lower)
małe_l = [m.lower() for m in text]
print(małe_l)

# Zamień spacje na podkreślniki (replace)
podkreśl = [x.replace(" ","_") for x in text]
print(podkreśl)
```

```python
# Masz string CSV: "Alice,28,Data Scientist,75000" . Użyj split(',') aby wyciągnąć:
# Imię
# Wiek (jako int)
# Zawód
# Pensję (jako int)
# Wyświetl te informacje w formacie: "Alice ma 28 lat, pracuje jako Data Scientist i
# zarabia $75000"

zdanie = "Alice,28,Data Scientist,75000"
x = zdanie.split(',')
Imię, Wiek, Zawód, Pensja = x
print(f'{Imię} ma {Wiek} lat, pracuje jako {Zawód} i zarabia ${Pensja}.')
```
