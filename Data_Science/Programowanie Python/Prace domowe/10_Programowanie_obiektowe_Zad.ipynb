```python
# Stwórz 4 komponenty:
# FiltrowanieDodatnich – usuwa liczby ujemne
# Zaokraglanie – zaokrągla do całkowitych
# Skalowanie – mnoży przez stały współczynnik
# Sumowanie – zwraca sumę listy
# Stwórz klasę Pipeline i przetestuj na danych: [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]

class Pipeline:
    def FiltrowanieDodatnich(self, dane):
        return [x for x in dane if x >= 0]
    def Zaokraglenie(self, dane):
        return [round(x) for x in dane]
    def Skalowanie(self, dane, stala):
        return [x * stala for x in dane]
    def Sumowanie(self, dane):
        return sum(dane)

dane = [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]
p = Pipeline()
print(f"Po filtrowaniu liczb dodatnich: {p.FiltrowanieDodatnich(dane)}")
print(f"Po zaokrągleniu: {p.Zaokraglenie(dane)}")
print(f"Po skalowaniu {p.Skalowanie(dane, 6)}")
print(f"Po sumowaniu: {p.Sumowanie(dane)}")
```

```python
# Stwórz klasę Produkt (nazwa, cena) oraz klasę Zamowienie , która zawiera listę
# produktów.
# Metody Zamowienie :
# dodaj_produkt(produkt, ilosc)
# laczna_wartosc() – suma wartości zamówienia
# pokaz_zamowienie() – wyświetla wszystkie produkty i łączną wartość
# Stwórz zamówienie z 5 produktami

class Produkt:
    def __init__(self, nazwa, cena):
        self.nazwa = nazwa
        self.cena = cena
class Zamowienie:
    def __init__(self):
        self.produkt = []
    def dodaj_produkt(self, produkt, ilosc):
        self.produkt.append((produkt,ilosc))
    def laczna_wartosc(self):
        suma = 0
        for produkt, ilosc in self.produkt:
            suma += produkt.cena * ilosc
        return suma
    def pokaz_zamowienie(self):
        print(f"Łączna wartość zamówienia wynosi: {self.laczna_wartosc()} zł.")

prod1 = Produkt("Monitor", 1000)
prod2 = Produkt("Myszka", 30)
prod3 = Produkt("Klawiatura", 300)
prod4 = Produkt("Lampka", 160)
prod5 = Produkt("Laptop", 3000)

zam = Zamowienie()
zam.dodaj_produkt(prod1, 4)
zam.dodaj_produkt(prod2, 3)
zam.dodaj_produkt(prod3, 8)
zam.dodaj_produkt(prod4, 1)
zam.dodaj_produkt(prod5, 7)

zam.pokaz_zamowienie()
```

```python
# Stwórz klasę bazową Figura z metodą pole() .
# Stwórz klasy potomne:
# Prostokat(szerokosc, wysokosc)
# Kolo(promien)
# Każda klasa implementuje pole() .
# Stwórz listę figur i oblicz łączne pole
class Figura: 
    def pole(self):
        pass
class Kolo(Figura):
    def __init__(self, promien):
        self.promien = promien
    def pole(self):
        return 3.14 * (self.promien ** 2)
class Prostokat(Figura):
    def __init__(self, szerokosc, wysokosc):
        self.szerokosc = szerokosc
        self.wysokosc = wysokosc
    def pole(self):
        return self.szerokosc * self.wysokosc
    
figury = [
    Kolo(5),
    Prostokat(12, 5),
    Kolo(9),
    Prostokat(2, 6)
]

suma = 0
for f in figury:
    suma += f.pole()

print(f"Suma wynosi: {suma:.3f}")
```
