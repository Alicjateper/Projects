```python
# Wyczyść dataset Titanic stosując pełny pipeline:
# Wymagania:
# Usuń kolumny z >60% NaN
# Wypełnij Age medianą
# Wypełnij Embarked dominantą
# Usuń wiersze z pozostałymi NaN
# Przed i po: wypisz kształt DataFrame
# Sprawdź czy nie ma NaN
# Porównaj rozkład Age przed i po
import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

titanic = pd.read_csv(url)
print(titanic)

threshold = len(titanic) * 0.6
titanic = titanic.dropna(thresh=threshold,axis=1)

kopia = titanic.copy()
titanic_mediana = kopia['Age'].median()
kopia['Age'] = kopia['Age'].fillna(titanic_mediana)

embarked_mode = kopia['Embarked'].mode()[0] 
kopia['Embarked'] = kopia['Embarked'].fillna(embarked_mode)
print(f"\nEmbarked po wypełnieniu dominantą ('{embarked_mode}'):")
print(kopia['Embarked'].value_counts())

kopia.dropna(axis='rows').head()
```

```python
# Stwórz nowe zmienne z Titanic:
# Wymagania:
# FamilySize = SibSp + Parch + 1
# IsAlone = 1 jeśli FamilySize == 1, inaczej 0
# AgeGroup = kategorie (0-12, 13-18, 19-35, 36-60, 60+)
# Dodaj wszystkie 3 zmienne do DataFrame
# Wyświetl value_counts() dla każdej
# Oblicz wskaźnik przeżycia dla IsAlone vs nie alone

import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

titanic = pd.read_csv(url)
titanic_clean = titanic.copy()
print(titanic_clean)

titanic_clean['FamilySize'] = titanic_clean['SibSp'] + titanic_clean['Parch'] + 1
print(titanic_clean['FamilySize'].value_counts().head())

titanic_clean['IsAlone'] = (titanic_clean['FamilySize'] == 1).astype(int)
print(f"\nPasażerowie samotni: {titanic_clean['IsAlone'].sum()}")

przedzial_wiekowy = [0, 12, 18, 35, 60, 120]
kat = ['Dziecko', 'Nastolatek', 'Dorosły', 'Starszy', 'Senior']
titanic_clean["AgeGroup"] = pd.cut(titanic_clean['Age'], bins=przedzial_wiekowy, labels=kat)
print(titanic_clean['AgeGroup'].value_counts())
```
