```python
# Utwórz tablicę NumPy zawierającą liczby od 1 do 100. Oblicz i wypisz:
# Wymagania:
# Sumę wszystkich elementów
# Średnią arytmetyczną
# Medianę
# Wartość maksymalną i minimalną
# Użyj np.arange() do stworzenia tablicy
# Użyj funkcji agregujących NumPy ( np.sum , np.mean , etc.)

import numpy as np

tablica = np.arange(1,101)
print(f"Tablica: {tablica}")
print(f"Suma wszystkich elementów: {np.sum(tablica)}")
print(f"Średnia arytmetyczna: {np.mean}")
print(f"Mediana: {np.median(tablica)}")
print(f"Wartość maksymalna: {np.max(tablica)}, wartość minimalna: {np.min(tablica)}")
```

```python
# Masz dwie tablice reprezentujące ceny produktów przed i po rabacie:
# Oblicz i wypisz:
# before = np.array([100, 250, 75, 320, 180, 450, 90])
# after = np.array([80, 200, 60, 280, 150, 400, 75])
# Wymagania:
# Różnicę między cenami (oszczędności)
# Procent rabatu dla każdego produktu
# Średnią oszczędność
# # Użyj operacji wektorowych (bez pętli)
# Wynik zaokrąglij do 2 miejsc po przecinku

import numpy as np

before = np.array([100, 250, 75, 320, 180, 450, 90])
print(f"Before: {before}")
after = np.array([80, 200, 60, 280, 150, 400, 75])
print(f"After: {after}")
print(f"Różnica między cenami: {before - after}")
print(f"Procent rabatu: {np.round((100 - ((after * 100) / before)), 2)}")
```

```python
# Dana tablica 1D:
# # Wykonaj:
# temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])
# Wypisz temperatury z dni parzystych (0, 2, 4, ...)
# Wypisz ostatnie 5 temperatur
# Wypisz temperatury w odwrotnej kolejności
# Znajdź temperatury większe niż 20

import numpy as np

temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])
print("Temperatura w dni parzyste:", temps[::2])
print("Temperatura w odwrotnej kolejności:", temps[::-1])
x = temps[temps >= 20]
print("Temperatury wyższe niż 20", x)
```

```python
# Dana tablica ocen (5 studentów x 4 egzaminy):
# Dodaj bonus do każdego egzaminu: [5, 10, 5, 10]
# Następnie pomnóż wyniki każdego studenta przez mnożnik: [1.0, 1.1, 0.95, 1.15, 1.0]
# Wymagania:
# Użyj broadcastingu (bez pętli)
# Wypisz wyniki przed i po
import numpy as np

exams = np.array([
[78, 85, 92, 88],
[65, 72, 68, 70],
[90, 88, 95, 92],
[55, 60, 58, 62],
[82, 79, 85, 80]
])
print("Wyniki egzaminów:")
print(exams)
bonus = np.array([5, 10, 5, 10])
print("Bonus:", bonus)
bonus_egzamin = exams + bonus
print("Wyniki egzaminów z bonusem:")
print(bonus_egzamin)
mnoznik = np.array([1.0, 1.1, 0.95, 1.15, 1.0]).reshape(5,1)
print("Monożnik:")
print(mnoznik)
wynik_mnoznika = exams * mnoznik
print("Po przemnożeniu:")
print(wynik_mnoznika)
```
