```python
results = []
accuracy = [0.75, 0.82, 0.79, 0.85, 0.88]
results.extend(accuracy)
# Wyświetl długość listy
# Wyświetl średnią wartość
# Wyświetl maksymalną wartość
print(f"Dodanie do listy, {results} ")

resualts2 = len(results)
print(f"Długość listy wynosi: {resualts2}")
średnia = sum(results) / resualts2
print(f"Średnia wynosi: {średnia}")
max = max(results)
print(f"Maxymalna watrość wynosi: {max}")
```

```python
text = [" Customer ID ", "TOTAL AMOUNT", " date "]
# Usuń białe znaki (strip)
delete = []
for t in text:
    t = t.strip()
    delete.append(t)
print(delete)
# Zamień na małe litery (lower)
małe_l = [m.lower() for m in text]
print(małe_l)

# Zamień spacje na podkreślniki (replace)
podkreśl = [x.replace(" ","_") for x in text]
print(podkreśl)
```

```python
# Masz listę transakcji (kategoria, kwota):
# Oblicz łączną kwotę dla każdej kategorii. Wynik zapisz w słowniku (użyj pętli i warunków)

transactions = [
('food', 50),
('transport', 20),
('food', 30),
('entertainment', 40),
('transport', 15),
('food', 25)
]
suma_category = {}
for category, amount in transactions:
    # if category not in suma_category:
    #     suma_category[category] = 0
    # suma_category[category] += amount
    suma_category[category] = suma_category.get(category,0) + amount
print(f"Suma wynosi: {suma_category}")
```

```python
# Masz listę pomiarów: [23, 24, 22, 150, 23, 25, -10, 24, 23] . Outlierem jest wartość
# poza zakresem [20, 30]. Stwórz nową listę bez outlierów używając list comprehension

Pomiary = [23, 24, 22, 150, 23, 25, -10, 24, 23] 
poza_zakresem = [x for x in Pomiary if x < 20 or x > 30]
print(f"Nowa lista: {poza_zakresem}")
```
