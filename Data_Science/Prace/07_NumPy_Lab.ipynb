```python
# Utwórz tablicę NumPy zawierającą liczby od 1 do 100. Oblicz i wypisz:
# Wymagania:
# Sumę wszystkich elementów
# Średnią arytmetyczną
# Medianę
# Wartość maksymalną i minimalną
# Użyj np.arange() do stworzenia tablicy
# Użyj funkcji agregujących NumPy ( np.sum , np.mean , etc.)

import numpy as np

tablica = np.arange(1,101)
print(f"Tablica: {tablica}")
print(f"Suma wszystkich elementów: {np.sum(tablica)}")
print(f"Średnia arytmetyczna: {np.mean}")
print(f"Mediana: {np.median(tablica)}")
print(f"Wartość maksymalna: {np.max(tablica)}, wartość minimalna: {np.min(tablica)}")
```

```python
# Masz dwie tablice reprezentujące ceny produktów przed i po rabacie:
# Oblicz i wypisz:
# before = np.array([100, 250, 75, 320, 180, 450, 90])
# after = np.array([80, 200, 60, 280, 150, 400, 75])
# Wymagania:
# Różnicę między cenami (oszczędności)
# Procent rabatu dla każdego produktu
# Średnią oszczędność
# # Użyj operacji wektorowych (bez pętli)
# Wynik zaokrąglij do 2 miejsc po przecinku

import numpy as np

before = np.array([100, 250, 75, 320, 180, 450, 90])
print(f"Before: {before}")
after = np.array([80, 200, 60, 280, 150, 400, 75])
print(f"After: {after}")
print(f"Różnica między cenami: {before - after}")
print(f"Procent rabatu: {np.round((100 - ((after * 100) / before)), 2)}")
```

```python
# Dana tablica 1D:
# # Wykonaj:
# temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])
# Wypisz temperatury z dni parzystych (0, 2, 4, ...)
# Wypisz ostatnie 5 temperatur
# Wypisz temperatury w odwrotnej kolejności
# Znajdź temperatury większe niż 20

import numpy as np

temps = np.array([15, 18, 22, 25, 23, 20, 17, 14, 16, 19, 21, 24])
print("Temperatura w dni parzyste:", temps[::2])
print("Temperatura w odwrotnej kolejności:", temps[::-1])
x = temps[temps >= 20]
print("Temperatury wyższe niż 20", x)
```

```python
# Dana tablica reprezentująca sprzedaż 4 produktów w 7 dni:
# sales = np.random.randint(10, 100, size=(4, 7))
# Oblicz i wypisz:
# Średnią sprzedaż per produkt (axis=1)
# Średnią sprzedaż per dzień (axis=0)
# Dzień z najwyższą sprzedażą (np.argmax)
# Produkt z najniższą średnią sprzedażą

import numpy as np

sales = np.random.randint(10, 100, size=(4, 7))
print(sales)
srednia_sales = np.mean((sales), axis=1)
print("Średnia sprzedaż per produkt:", srednia_sales)
srednia_dzien = np.mean((sales), axis=0)
print("Średnia sprzedaż per dzień:", srednia_dzien)
day1 = np.sum(sales[: ,0])
day2 = np.sum(sales[: ,1])
day3 = np.sum(sales[: ,2])
day4 = np.sum(sales[: ,3])
day5 = np.sum(sales[: ,4])
day6 = np.sum(sales[: ,5])
day7 = np.sum(sales[: ,6])

days = ([
    [day1],
    [day2],
    [day3],
    [day4],
    [day5],
    [day6],
    [day7]
])
high_sales = np.argmax(days)
print("Najwyższa sprzedaś w dniu", high_sales)
```

```python
# Masz tablicę z oceną studentów:
# Wykonaj:
# grades = np.array([78, 45, 92, 67, 34, 88, 56, 91, 23, 85, 73, 95])
# Znajdź wszystkie oceny >= 80 (oceny wysokie)
# Znajdź wszystkie oceny < 50 (oceny niedostateczne)
# Policz ile jest ocen w przedziale [60, 80)
# Zmień wszystkie oceny < 50 na 50 (podniesienie do minimum)

import numpy as np

grades = np.array([78, 45, 92, 67, 34, 88, 56, 91, 23, 85, 73, 95])
oceny_wysokie = grades[grades >= 80] 
print("Wysokie oceny:", oceny_wysokie)
oceny_niedostateczny = grades[grades < 50]
print("Oceny niedostateczny:", oceny_niedostateczny)
oceny_przedzial = np.sum((grades >= 60) & (grades <= 80))
print("Ilość ocen w przedziale (60, 80):", oceny_przedzial)
```

```python
# Dana tablica ocen (5 studentów x 4 egzaminy):
# Dodaj bonus do każdego egzaminu: [5, 10, 5, 10]
# Następnie pomnóż wyniki każdego studenta przez mnożnik: [1.0, 1.1, 0.95, 1.15, 1.0]
# Wymagania:
# Użyj broadcastingu (bez pętli)
# Wypisz wyniki przed i po
import numpy as np

exams = np.array([
[78, 85, 92, 88],
[65, 72, 68, 70],
[90, 88, 95, 92],
[55, 60, 58, 62],
[82, 79, 85, 80]
])
print("Wyniki egzaminów:")
print(exams)

bonus = np.array([5, 10, 5, 10])
print("Bonus:", bonus)

bonus_egzamin = exams + bonus
print("Wyniki egzaminów z bonusem:")
print(bonus_egzamin)

mnoznik = np.array([1.0, 1.1, 0.95, 1.15, 1.0]).reshape(5,1)
print("Monożnik:")
print(mnoznik)

wynik_mnoznika = exams * mnoznik
print("Po przemnożeniu:")
print(wynik_mnoznika)
```

```python
# Dataset: Sprzedaż 6 produktów w 30 dni
# Oblicz:
# sales = np.random.randint(50, 200, size=(6, 30))
# Trend sprzedaży per produkt (różnica między ostatnimi 7 dniami a pierwszymi 7 dniami)
# Najlepszy tydzień (dni 0-6, 7-13, 14-20, 21-27) dla całej sprzedaży
# Wykres: sprzedaż per produkt w czasie (linie)
# Użyj slicingu do wycinania tygodni
# Wizualizacja matplotlib
import numpy as np
import matplotlib.pyplot as plt

sales = np.random.randint(50, 200, size=(6, 30))
print("sales:")
print(sales)

last_week = np.sum(sales[:,-7:], axis=1)
print("ostatnie:")
print(last_week)

first_week = np.sum(sales[:,0:7],axis=1)
print("Pierwsze:")
print(first_week)

roznica = last_week - first_week
print("Różnica:")
print(roznica)

week1 = np.sum(sales[:, 0:7])
week2 = np.sum(sales[:, 7:14])
week3 = np.sum(sales[:, 14:21])
week4 = np.sum(sales[:, 21:28])

weeks = np.array([
    [week1],
    [week2],
    [week3],
    [week4],
])

naj_tydz = np.argmax(weeks)
print("Najlepszy tydzień to tydzień numer:", naj_tydz + 1)

```
