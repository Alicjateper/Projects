```python
# Wczytaj dataset Titanic i oblicz średni wiek pasażerów według klasy ( Pclass ) i płci ( Sex ).
# Dataset: Titanic
# (https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv)
# Wymagania:
# Użyj .groupby() z dwoma kolumnami
# Oblicz średnią wieku ( .mean() )
# Wyświetl wyniki w formie czytelnej tabeli
# Tabela ze średnim wiekiem dla 6 kombinacji (3 klasy × 2 płcie)

import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

titanic = pd.read_csv(url)

srednia = titanic.groupby(['Pclass','Sex'])['Age'].mean().unstack()
print(srednia)


```

```python
# Napisz własną funkcję, która oblicza zakres międzykwartylowy (IQR = Q3 - Q1) i zastosuj ją
# do grupy.
# Dataset: California Housing
# Wymagania:
# Napisz funkcję iqr_func(x) obliczającą IQR
# Zastosuj ją do MedHouseVal według regionów
# Porównaj IQR między regionami
# Tabela z IQR dla każdego regionu

from sklearn.datasets import fetch_california_housing
import pandas as pd

# pobranie datasetu
data = fetch_california_housing()
frame = pd.DataFrame(data.data, columns=data.feature_names)

def iqr_func(x):
    Q1 = x.quantile(0.25)
    Q3 = x.quantile(0.75)
    return Q3 - Q1

# dodanie zmiennej docelowej
frame["MedHouseVal"] = data.target

# stworzenie regionów (np. na podstawie Latitude)
frame["Region"] = pd.cut(
    frame["Latitude"],
    bins=3,
    labels=["South", "Central", "North"]
)

# zastosowanie funkcji IQR do grup
result = frame.groupby("Region")["MedHouseVal"].transform(iqr_func)

# wyświetlenie tabeli
print(result)

```

```python
# Dla datasetu California Housing znormalizuj MedInc (dochód) w każdym regionie do
# zakresu [0, 1].
# Dataset: California Housing
# Wymagania:
# Stwórz regiony (np. na podstawie Latitude > median)
# Napisz funkcję normalizacji Min-Max
# Użyj .groupby().transform() lub .apply()
# Zweryfikuj, że min=0 i max=1 w każdym regionie
# DataFrame z kolumną znormalizowaną grupowo

from sklearn.datasets import fetch_california_housing
import pandas as pd

# wczytanie danych
data = fetch_california_housing()

frame = pd.DataFrame(data.data, columns=data.feature_names)
frame["MedHouseVal"] = data.target

# 1. tworzenie regionów (Latitude > median)
median_lat = frame["Latitude"].median()

frame["Region"] = (frame["Latitude"] > median_lat).map(
    {True: "North", False: "South"}
)

# 2. funkcja Min-Max
def min_max(x):
    return (x - x.min()) / (x.max() - x.min())

# 3. normalizacja grupowa MedInc
frame["MedInc_norm"] = frame.groupby("Region")["MedInc"].transform(min_max)

# 4. weryfikacja 
check = frame.groupby("Region")["MedInc_norm"].agg(["min", "max"])
print(check)

# 5. wynikowy DataFrame
print(frame[["Region", "MedInc", "MedInc_norm"]].head())
```
