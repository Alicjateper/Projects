```python
# Stwórz dwa wektory 2D: v1 = [3, 4] i v2 = [1, 2]. Oblicz ich sumę, różnicę oraz narysuj
# wszystkie trzy wektory na wykresie używając plt.quiver() .
# Stwórz wektory jako NumPy arrays
# Oblicz v1 + v2 i v1 - v2
# Narysuj wszystkie 4 wektory (v1, v2, suma, różnica) różnymi kolorami
# Dodaj legendę i grid
# Wykres z 4 strzałkami reprezentującymi wektory

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 8))

origin = np.array([0, 0])

# Stwórz wektory jako NumPy arrays
v1 = np.array([3,4])
print("Wektor v1:\n", v1)

v2 = np.array([1,2])
print("Wektor v2:\n", v2)

# Oblicz v1 + v2 i v1 - v2
suma = v1 + v2
print("Suma wektorów wynosi:\n", suma)

roznica = v1 - v2
print("Różnica wektorów wynosi:\n", roznica)

# Narysuj wszystkie 4 wektory (v1, v2, suma, różnica) różnymi kolorami
ax.quiver(*origin, *v1, angles='xy', scale_units='xy', scale=1, color='blue', width=0.006, label='v1 = [3, 4]')
ax.quiver(*origin, *v2, angles='xy', scale_units='xy', scale=1, color='red', width=0.006, label='v2 = [1, 2]')
ax.quiver(*origin, *suma, angles='xy', scale_units='xy', scale=1, color='green', width=0.006, label='v1 + v2 = [4, 6]')
ax.quiver(*origin, *roznica, angles='xy', scale_units='xy', scale=1, color='pink', width=0.006,label='v1 - v2 = [2, 2]')

# Równoległobok dodawania wektorów (wizualizacja sumy)
ax.quiver(*v1, *v2, angles='xy', scale_units='xy', scale=1, color='red', width=0.004, alpha=0.3, linestyle='dashed')
ax.quiver(*v2, *v1, angles='xy', scale_units='xy', scale=1, color='blue', width=0.004, alpha=0.3, linestyle='dashed')

ax.set_xlim(-1, 10)
ax.set_ylim(-1, 9)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)

ax.set_title('Dodawanie wektorów w przestrzeni 2D', fontsize=14)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```

```python
# Wczytaj cały dataset Iris (4 cechy). Zaimplementuj standaryzację (mean=0, std=1) dla
# każdej cechy używając operacji NumPy. Porównaj wynik z StandardScaler z sklearn.
# Wymagania:
# Iris (sklearn.datasets.load_iris)
# Oblicz mean i std per kolumna ( axis=0 )
# Zastosuj wzór: (X - mean) / std
# Sprawdź mean i std po standaryzacji
# Porównaj z sklearn.preprocessing.StandardScaler
# Oblicz max różnicę między implementacjami
# -> Macierz X_standardized (150, 4)
# -> Weryfikacja: mean≈0, std≈1
# -> Max różnica vs sklearn

import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data # Macierz cech (150, 4)
print("iris.data:\n", X)

# Standaryzacja: (X - mean) / std - dla każdej kolumny
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

print("Mean:", mean)
print("Std:", std)

standardized = (X - mean) / std

# Weryfikacja: mean≈0, std≈1
print(f"Po standaryzacji:\n")
print(f" Średnia per cecha: {standardized.mean(axis=0)}")
print(f" Std per cecha: {standardized.std(axis=0)}")

# Max różnica vs sklearn
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_sklearn = scaler.fit_transform(X)
print(f" Max różnica vs sklearn wynosi: {np.abs(standardized - X_sklearn).max():.10f}") #abs liczy wartość bezwględną
```

```python
# Wybierz 10 losowych obserwacji z Iris. Oblicz pełną macierz odległości euklidesowych
# (10×10) między wszystkimi parami. Wizualizuj jako heatmapę.
# Dataset: Iris (sklearn.datasets.load_iris)
# Wymagania: Losowo wybierz 10 obserwacji (użyj np.random.choice na indeksach)
# Użyj euclidean_distances z sklearn lub zagnieżdżone pętle
# Wizualizuj macierz jako heatmapę (seaborn.heatmap lub plt.imshow)
# Diagonala powinna być 0 (odległość od samego siebie)
# Oczekiwany rezultat: Macierz 10×10
# Heatmapa z annotacjami

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.metrics.pairwise import euclidean_distances

# Wczytywanie danych
iris = load_iris()
X = iris.data

# Wybranie 10 losowych obserwacji
np.random.seed(42)  # dla powtarzalności
indices = np.random.choice(X.shape[0], size=10, replace=False)
X_sample = X[indices]

# Obliczanie macierz odległości (10x10)
dist_matrix = euclidean_distances(X_sample, X_sample)

# Sprawdzenie diagonali
print("Diagonal:", np.diag(dist_matrix))  #sprawdzenie wartości na przekątnej macierzy

# Wizualizacja heatmapy
plt.figure(figsize=(8, 6))

plt.imshow(dist_matrix, cmap="viridis") #zamiast seaborn.heatmap
plt.colorbar()

plt.title("Macierz odległości euklidesowych (10x10)")
plt.xlabel("Obserwacja")
plt.ylabel("Obserwacja")

plt.show()
```
