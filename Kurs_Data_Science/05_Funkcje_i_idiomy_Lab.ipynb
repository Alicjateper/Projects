```python
# Napisz funkcję normalize_minmax(dane) , która normalizuje listę liczb do zakresu [0, 1]
# używając metody min-max.
# Wymagania:
# Wzór: (x - min) / (max - min)
# Funkcja zwraca nową listę znormalizowanych wartości
# Zaokrąglij do 3 miejsc po przecinku


wynagrodzenia = [3500, 4200, 5800, 7200, 9500, 4800, 6100]
def dane(wynagrodzenia):
    a = min(wynagrodzenia)
    b = max(wynagrodzenia)
    normalize_minmax = []
    for wartosc in wynagrodzenia:
        wartosc = round((wartosc - a) / (b - a),3)
        normalize_minmax.append(wartosc)
    return normalize_minmax
wynik = dane(wynagrodzenia)
print(wynik)
```

```python
# Masz dane klientów: wiek i dochód. Dodaj nowe cechy: kat_wiekowa (młody/
# średni/starszy) i kat_dochodu (niski/średni/wysoki).
# Wymagania:
# Kategoria wiekowa: <30 = młody, 30-50 = średni, >50 = starszy
# Kategoria dochodu: <5000 = niski, 5000-10000 = średni, >10000 = wysoki
# Użyj list comprehension do tworzenia nowej listy z dodatkowymi cecham

klienci = [
{"id": 1, "wiek": 25, "dochod": 4500},
{"id": 2, "wiek": 45, "dochod": 9200},
{"id": 3, "wiek": 32, "dochod": 6800},
{"id": 4, "wiek": 28, "dochod": 5200},
{"id": 5, "wiek": 55, "dochod": 12000}
]



dla_wieku = ["młody" if x["wiek"] < 30 else "średni" if 30 < x["wiek"] < 50 else "starszy"
    for x in klienci
    ]
dla_dochodu = ["niski" if x["dochod"] < 5000 else "średni" if 5000 < x["dochod"] <10000 else "wysoki"
    for x in klienci
    ]
nowa_lista = []
for x, a, b in zip(klienci, dla_wieku, dla_dochodu):
    nowa_lista.append({ 
        "id": x["id"],
        "wiek": x["wiek"],
        "dochód": x["dochod"],
        "kategoria_wiekowa": a,
        "kategoria_dochodu": b
     })
for klient in nowa_lista:
    print(klient) 
   
```

```python
# Napisz funkcję statystyki(liczby) , która przyjmuje listę liczb i zwraca tuple zawierający:
# (suma, średnia, minimum, maksimum).
# Dataset: Lista ocen: [3.5, 4.0, 5.0, 3.0, 4.5, 5.0, 4.0, 3.5]
# Wymagania:
# Funkcja powinna zwracać tuple z 4 wartościami
# Średnia zaokrąglona do 2 miejsc po przecinku
# -> Wypisanie wszystkich statystyk na ekran

def func (lista):
    suma = sum(lista)
    średnia = suma / len(lista)
    minimum = min(lista)
    maxium = max(lista)
    return suma, średnia, minimum, max

lista_testowa = [3.5, 4.0, 5.0, 3.0, 4.5, 5.0, 4.0, 3.5]
su, sr, mi, mx = func(lista_testowa)
print(su, sr, mi, mx)
```

```python
# Napisz trzy funkcje: celsius_to_fahrenheit(c) , fahrenheit_to_celsius(f) ,
# celsius_to_kelvin(c) .
# Dataset: Temperatury w Celsius: [0, 10, 20, 30, 37, 100]
# Wymagania:
# Każda funkcja zwraca skonwertowaną wartość
# Zastosuj funkcje do konwersji wszystkich temperatur
# -> Tabela z trzema kolumnami: °C, °F, K

def celsius_to_fahrenheit(c):
    return c * 9/5 + 32


def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9


def celsius_to_kelvin(c):
    return c + 273.15


temperatura = [0, 10, 20, 30, 37, 100]

lista_temp = [
    (
        temp,
        celsius_to_fahrenheit(temp),
        celsius_to_kelvin(temp)
    )
    for temp in temperatura
]

print(lista_temp)
```

```python
# Masz listę tuple z produktami: (nazwa, cena, ocena) . Posortuj ją według ceny rosnąco,
# używając sorted() i lambda.
# Dataset:
# produkty = [
# ("laptop", 2999, 4.5),
# ("mysz", 89, 4.2),
# ("klawiatura", 249, 4.8),
# ("monitor", 899, 4.3)
# ]
# Wymagania:
# Użyj sorted() z parametrem key=lambda ...
# Posortuj według drugiego elementu tuple (cena)
# -> Lista posortowana od najtańszego do najdroższego

produkty = [
("laptop", 2999, 4.5),
("mysz", 89, 4.2),
("klawiatura", 249, 4.8),
("monitor", 899, 4.3)
]

posortowane = sorted(produkty, key=lambda  x: x[1])
print(posortowane)
for produkty in posortowane:
    print(f" Produkt {produkty[0]}, cena: {produkty[1]}, ocena: {produkty[2]}")
```

```python
# Napisz funkcję normalize_minmax(dane) , która normalizuje listę liczb do zakresu [0, 1]
# używając metody min-max.
# Wymagania:
# Wzór: (x - min) / (max - min)
# Funkcja zwraca nową listę znormalizowanych wartości
# Zaokrąglij do 3 miejsc po przecinku


dane = [3500, 4200, 5800, 7200, 9500, 4800, 6100]

def normalize(dane):
    a = min(dane)
    b = max(dane)

    normalize_minmax = []

    for wartosc in dane:
        wartosc = round((wartosc - a) / (b - a),3)
        normalize_minmax.append(wartosc)
    return normalize_minmax
wynik = normalize(dane)
print(wynik)
```

```python
# Masz listę transakcji: [{"produkt": "X", "wartosc": 100}, ...] . Używając dict
# comprehension i grupowania, oblicz łączną wartość dla każdego produktu.
# Wymagania:
# Pogrupuj po produkcie
# Zsumuj wartości dla każdego produktu
# Wynik jako słownik: {"laptop": 5998, ...}

transakcje = [
{"produkt": "laptop", "wartosc": 2999},
{"produkt": "mysz", "wartosc": 89},
{"produkt": "laptop", "wartosc": 2999},
{"produkt": "klawiatura", "wartosc": 249},
{"produkt": "mysz", "wartosc": 89},
{"produkt": "mysz", "wartosc": 89}
]

wartosc_transakcji = {}
for tran in transakcje:
    prod = tran["produkt"]
    war = tran["wartosc"]
    if prod in wartosc_transakcji:
        wartosc_transakcji[prod] += war
    else:
        wartosc_transakcji[prod] = war
for produkt, wartosc in wartosc_transakcji.items():
    print(wartosc_transakcji)
```
